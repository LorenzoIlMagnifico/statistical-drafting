```python
# Get all available sets & modes for training. 
set_abbreviations, draft_modes = [], []
for fn in os.listdir("../data/17lands/"):
    if len(fn) > 20:
        sa, dm = fn.split(".")[1], fn.split(".")[2][:-5]
        set_abbreviations.append(sa)
        draft_modes.append(dm)

# Customize sets and modes for this training run. 
draft_modes = ["Premier"]
set_abbreviations = ["ECL"]

# Train models. 
for set_abbreviation, draft_mode in zip(set_abbreviations, draft_modes):
        try:
            print(f"Starting training for {set_abbreviation}, {draft_mode}")
            sd.default_training_pipeline(
                set_abbreviation,
                draft_mode,
                overwrite_dataset=True,
                top_drafters_only=True,
            )
        except Exception as e:
            print(f"Error for: {set_abbreviation}, {draft_mode}: {e}")
```
